```python
@latexify.function(use_math_symbols=True)
def konzentration(m_BTEX, m_MeOH, c_BTEX):
    return m_BTEX / (m_BTEX + m_MeOH) * c_BTEX

m_MeOH = Q_(6.3158, 'g')
m_btex = Q_(0.6060, 'g')
c_btex = Q_(200, 'ng/ul')

d = {
    r'm\_BTEX': m_btex,
    r'm\_MeOH': m_MeOH,
    r'c\_BTEX': c_btex
}

result = konzentration(m_btex, m_MeOH, c_btex)
print(f'Die Konzentration des Standards beträgt: {result}')

latexify_expression(str(konzentration), file_name='kal_btex_std', identifier=r'\mathrm{c_{Std}}', variables=d, result=result, result_idf=['cstd', 4])

# write_variables(result, 'cstd', precision=5)
write_variables(m_MeOH, 'mmeoh')
write_variables(m_btex, 'mbtex')
write_variables(c_btex, 'btexref')
```
