#### Aufruf geschieht über chemfig(), das erste Argument in Klammer ist der Name des Moleküls, das zweite der Name der entstehenden Datei

```python
chemfig('Coffein', file_name='molecule')
```

Auf PubChem: https://pubchem.ncbi.nlm.nih.gov/ sind die identifier (Compound CID) aller registrierten Moleküle zu finden. Für Coffein: 2519

```python
chemfig(2519, file_name='molecule')
```

##### Als weiteres Argument kann rotate= verwendet werden, damit rotiert das Molekül um die angegebene Gradzahl

```python
chemfig(2519, file_name='molecule', rotate=360, cmp_label='Coffein')
```
